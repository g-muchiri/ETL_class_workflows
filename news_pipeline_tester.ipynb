```python
import pandas as pd
import requests
import psycopg2
import sqlalchemy
```

```python
url = 'https://newsapi.org/v2/everything?q=apple&from=2026-06-23&to=2026-06-23&sortBy=popularity&apiKey=757a2dcfd035405b9fc631d9925f4d09'
```

```python
r = requests.get(url=url)
r.status_code
```

```python
## It returned the status code 200 to show that the URL is actually able to return data
print(type(r))
```

```python
test_data = r.json()
type(test_data)
```

```python
test_data
```

```python
articles_data = test_data['articles']
articles_data
```

```python
articles_df = pd.DataFrame(articles_data)
articles_df.head()
```

```python
## Frm the above data we do not need the column source
articles_df.drop(columns=['source'], inplace=True)
```

```python
## Being ready to load

## This depends on the database you want to install

## Below for example is a local database

## you could also use aiven

DATABASE_NAME = 'postgres'
DATABASE_USER = 'postgres'
DATABASE_HOST = 'localhost'
DATABASE_PORT = 5432
DATABASE_PASSWORD = 'root'

from sqlalchemy import create_engine, text

engine = create_engine(f'postgresql+psycopg2://{DATABASE_USER}:{DATABASE_PASSWORD}@{DATABASE_HOST}:{DATABASE_PORT}/{DATABASE_NAME}')
```

```python
## this is testing the connection
with engine.connect() as conn:
    result = conn.execute(text('SELECT 1'))
    for i in result:
        print(i)
```

```python
articles_df.to_sql('articles', engine, if_exists='replace', index=False)
```
